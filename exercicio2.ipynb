```python
numero = float(input("Digite o número da tabuada: "))
i = int(input("Digite o início do intervalo: "))
fim = int(input("Digite o fim do intervalo: "))

while i <= fim:
    print(numero, "x", i, "=", numero * i)
    i = i + 1
```

**QUESTÃO 3:**

```python
sexo = input('Digite o seu sexo: ')
controle = True
while controle:
  if sexo == 'F':
    controle = False
    print('Feminino')
  elif sexo == 'M':
    controle = False
    print('Masculino')
  else:
    print('Sexo Inválido')
    sexo = input('Digite o seu sexo: ')
```

**QUESTÃO 4:**

```python
distancia = float(input('Qual a distancia da viagem?(km): '))
valor_passagem = 0
if distancia > 0 and distancia < 200:
  valor_passagem = distancia * 0.50
else:
  valor_passagem = distancia * 0.45
print(f'O valor da passagem é de R${valor_passagem:.2f}')
```

**QUESTÃO 7:**

```python
palavra = input("Digite uma palavra: ")

vogais = 0
tem_a = False

for letra in palavra:
    print(letra.upper())

    if letra == "a" or letra == "e" or letra == "i" or letra == "o" or letra == "u":
        vogais = vogais + 1

    if letra == "a":
        tem_a = True

print(vogais)

if tem_a == True:
    print("A letra 'a' está presente na palavra.")
else:
    print("A letra 'a' não está presente na palavra.")
```
